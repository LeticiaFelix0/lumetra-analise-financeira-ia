Foram identificados 25 registros com desconto ausente. Não foi possível determinar a causa da ausência, não podemos simplesmente substituir todos os NaN de desconto por 0, porque perderíamos informação.. Para preservar esses registros, podemos reconstruir o desconto usando as outras informações da própria venda.

```python
descontos_nulos = df[df["desconto"].isnull()].copy()

descontos_nulos["desconto_calculado"] = (
    (descontos_nulos["quantidade"] * descontos_nulos["preco_unitario"])
    - descontos_nulos["faturamento"]
) / (
    descontos_nulos["quantidade"] * descontos_nulos["preco_unitario"]
)

descontos_nulos[
    ["produto", "quantidade", "preco_unitario",
     "faturamento", "desconto_calculado"]
]

descontos_nulos["desconto_percentual"] = (
    descontos_nulos["desconto_calculado"] * 100
).round(0)

descontos_nulos[
    ["produto", "quantidade", "preco_unitario",
     "faturamento", "desconto_percentual"]
]
```

```python
df[df["id_cliente"].isnull()]
```

Foram identificados 25 registros com id_cliente ausente. Não foi possível determinar a causa da ausência nem reconstruir o identificador a partir das demais informações disponíveis. Para preservar esses registros sem atribuir clientes fictícios, os valores ausentes serão classificados como Não informado.

```python
validacao = df[
    df["desconto"].notnull() &
    (df["quantidade"] > 0) &
    (df["preco_unitario"] > 0)
].copy()

validacao["faturamento_calculado"] = (
    validacao["quantidade"]
    * validacao["preco_unitario"]
    * (1 - validacao["desconto"])
).round(2)

validacao[
    validacao["faturamento"] != validacao["faturamento_calculado"]
][
    ["id_pedido", "quantidade", "preco_unitario",
     "desconto", "faturamento", "faturamento_calculado"]
]
```

```python
validacao["faturamento"] != validacao["faturamento_calculado"]
```

```python
validacao["diferenca_faturamento"] = (
    validacao["faturamento"] -
    validacao["faturamento_calculado"]
).abs()
```

```python
validacao["diferenca_faturamento"].describe()
```

Validação do faturamento: ao comparar o faturamento registrado com o faturamento recalculado a partir da quantidade, preço unitário e desconto, foram observadas pequenas diferenças. A maior divergência encontrada foi de R$ 0,01, indicando diferenças compatíveis com arredondamento, sem evidência de inconsistências relevantes no cálculo do faturamento entre os registros avaliados.

## Seção 11 — Análise Temporal

11.1 Criando a coluna de mês

```python
df_concluido["mes"] = (
    df_concluido["data_hora"].dt.to_period("M")
)
```

11.2 Faturamento mensal

```python
df_concluido[
    ["data_hora", "mes", "faturamento"]
].head(10)
```

```python
resumo_financeiro_mensal["variacao_faturamento"] = (
    resumo_financeiro_mensal["faturamento"]
    .pct_change()
    .mul(100)
    .round(2)
)
```

```python
resumo_financeiro_mensal[
    ["faturamento", "variacao_faturamento"]
]
```
